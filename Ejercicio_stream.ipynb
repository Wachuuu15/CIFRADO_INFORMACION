## Ejercicio Block Cipher

```python
import random
from typing import List
```

```python
def generate_keystream(seed: int, length: int) -> List[int]:
    random.seed(seed)  
    return [random.randint(0, 255) for _ in range(length)]  # Generar keystream

```

```python
def xor_encrypt(plaintext: str, keystream: List[int]) -> List[int]:
    plaintext_bytes = plaintext.encode('utf-8')
    return [b ^ k for b, k in zip(plaintext_bytes, keystream)]  

```

```python
def xor_decrypt(ciphertext: List[int], keystream: List[int]) -> str:
    decrypted_bytes = bytes([c ^ k for c, k in zip(ciphertext, keystream)])
    return decrypted_bytes.decode('utf-8')
```

```python
def test_xor_encryption():
    seed = 12345
    plaintext = "Hola, ¿qué tal?"
    plaintext_bytes = plaintext.encode('utf-8')
    keystream = generate_keystream(seed, len(plaintext_bytes))  # Generar keystream basado en los bytes reales

    # Cifrar
    ciphertext = xor_encrypt(plaintext, keystream)
    print("Texto Cifrado:", ciphertext)

    # Descifrar
    decrypted_text = xor_decrypt(ciphertext, keystream)
    print("Texto Descifrado:", decrypted_text)

    assert decrypted_text == plaintext, "El descifrado falló"
    print("Prueba pasada con éxito.")


test_xor_encryption()
```

### Preguntas

1. Cuando se cambia la clave utilizada para generar el keystream, se produce un keystream completamente diferente. Dado que el cifrado XOR es simétrico, si el keystream generado no coincide con el utilizado durante el cifrado, el descifrado fallará y se obtendrá un texto ilegible.

2. Reutilizar el mismo keystream para cifrar dos mensajes diferentes introduce una vulnerabilidad conocida como "ataque de texto plano conocido". Si un atacante obtiene ambos textos cifrados, puede realizar un análisis XOR entre ellos para descubrir patrones o recuperar parcialmente los mensajes originales.

3. La longitud del keystream debe ser al menos igual a la longitud del mensaje a cifrar. Si el keystream es más corto y se repite, se crean patrones predecibles que pueden ser explotados mediante análisis criptográfico. Un keystream suficientemente largo evita estas vulnerabilidades.

4. Al generar un keystream en un entorno real, se debe utilizar un generador de números aleatorios criptográficamente seguro. La clave debe ser única por cada mensaje cifrado para evitar reutilización. Además, el keystream debe tener la misma longitud que el mensaje y debe ser generado de manera impredecible para garantizar la seguridad.
